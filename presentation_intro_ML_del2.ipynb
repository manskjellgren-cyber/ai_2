#### Hantering av kategoriska variabler

Nominala – ingen ordning

Ordinala – tydlig ordning

OBS om en data en ordinal eller inte beror på situationen.

Skapar nya features. 

```python
from sklearn.preprocessing import OneHotEncoder

encoder = OneHotEncoder(drop="first", sparse_output=False, handle_unknown="ignore") # Ändra på drop för att byta mellan one-hot-encoding och dummy-variable-encoding. 

X = [["red"], ["blue"], ["green"], ["red"]]

X_encoded = encoder.fit_transform(X)

print(encoder.categories_)
print(X_encoded)
```

```python
from sklearn.preprocessing import OrdinalEncoder

categories = [["low", "medium", "high"]]

encoder = OrdinalEncoder(categories=categories)

X = [["low"], ["medium"], ["high"], ["medium"]]

X_encoded = encoder.fit_transform(X)

X_encoded
```

#### Hantering av outliers

En outlier är ett värde som avviker kraftigt från resten av datan och riskerar att dominera analys eller modell.

MEN alla outliers är inte fel. 

Åtgärder: 

* Felaktiga värden tas bort eller korrigeras.
* Transformera data. 
* Använd en mer robust modell. Trädmodeller påverkas generellt mindre.
