```python
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# 1. Läs in bilden
map_img = mpimg.imread('california.png')

# 2. Definiera bildens koordinatgränser (Exempelvärden, byt ut mot dina!)
# Format: [min_lon, max_lon, min_lat, max_lat]
img_extent = [-124.55, -113.80, 32.45, 42.05]

# 3. Skapa plotten
fig, ax = plt.subplots(figsize=(10, 8))

# Lägg in kartan som bakgrund
ax.imshow(map_img, extent=img_extent, alpha=0.5)

# 4. Plotta din dataframe ovanpå
# Se till att x är longitud och y är latitud
df.plot(kind='scatter', x='longitude', y='latitude', 
        ax=ax,                 # Mycket viktigt: plotta i samma axel
        c='median_house_value', # Exempel på färgstyrning
        cmap='Wistia', vmin= 1, vmax=1000000, 
        alpha=0.8)

plt.xlabel('Longitud')
plt.ylabel('Latitud')
plt.show()
```

ABOVE

This is the map over California

the Longitude(X-axis horizontal) and Latitude(Y-axis vertical) coordinates represents the WGS 84 standard used in GPS.
By creating a scatter of this GPS data aligned with the map image, a lot of coloured dots appears. Each dot on the map represent a "district" of California during the 90´s. The dots colour reveals the varying "median house value" of each district. (Green has lowest median value and Orange the highest)

 There are some issues to watch out for with this data. Actual buildings, size and prices are not present in the dataset. As a data analyst the following must be taken into consideration when performing the EDA:

Old data:
    Old data does not show todays house prices, nor how the architectural landscape has evolved. The housing.csv data was gathered some 30 years ago. It could give an indication but cannot be used professionally by a real estate broker.

Value statistics per district:
    The median house value of a district doesnt necessary represent the normal distribution of bouldings . as buildings in california varies from, manisons, hotels, skyscrapers, Entertainment parks, villas, residential areas, schools and trailor parks. And some district varies from a few citizens up to half a million.

Bias:
there is a lot of different buildings. Some buildings may not be representative and should therefore be excluded. 


As a data analyst it is necessary to discuss all of the above with the stakeholders to eliminate misinterpretations.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
# longitude	latitude	housing_median_age	total_rooms	total_bedrooms	population	households	median_income	median_house_value	ocean_proximity
df = pd.read_csv("housing.csv")
# Skapar en figur med x rader och y kolumner 
fig, axes = plt.subplots(nrows=7, ncols=1, figsize=(20, 30))
# Platta till arrayen för att enklare iterera eller komma åt dem (axes[0], axes[1]...)
ax = axes.flatten()
df.plot(x='ocean_proximity', y='housing_median_age', alpha=0.1, ax=ax[0], kind='scatter', title='Plot 1 housing_median_age')
df.plot(x='total_rooms', y='ocean_proximity', alpha=0.1, ax=ax[1], kind='scatter', title='Plot 2 total_rooms')
df.plot(x='total_bedrooms', y='ocean_proximity', alpha=0.1, ax=ax[2], kind='scatter', title='Plot 3 total_bedrooms')
df.plot(x='population', y='ocean_proximity', alpha=0.1, ax=ax[3], kind='scatter', title='Plot 4 population')
df.plot(x='households', y='ocean_proximity', alpha=0.1, ax=ax[4], kind='scatter', title='Plot 5 households')
df.plot(x='median_income', y='ocean_proximity', alpha=0.1, ax=ax[5], kind='scatter', title='Plot 6 median_income')
df.plot(x='median_house_value', y='ocean_proximity', alpha=0.1, ax=ax[6], kind='scatter', title='Plot 7 median_house_value')

#housing.plot(x='population', y='ocean_proximity', kind='scatter')
# Justera layouten så att titlar och etiketter inte krockar
plt.tight_layout()
plt.show()
```



ABOVE
Seven plofigures above shows the variable ocean_proximity - the only variable with datatype "string", plotted against all other relevant "float" variables from the housing.csv dataset.

There are five different district areas in the variable ocean_proximity:
    ISLAND
    NEAR OCEAN
    INLAND
    <1H OCEAN
    NEAR BAY

Looking at the seven plotted figures above proofs that ISLAND has very few districts and thus very few observations. 

A data analyst should therefore consider removing the district area ISLAND as it risks making the calculation more complex and also risks misguiding paths to proper suggestions.
